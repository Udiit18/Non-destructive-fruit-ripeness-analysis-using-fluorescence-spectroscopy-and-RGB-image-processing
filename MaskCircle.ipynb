```python
import cv2
import numpy as np

# Load image
image = cv2.imread(r"Data\Jujube\Sample_Images\B1_S2.jpeg")
# Chnage images in the above line line 
if image is None:
    print("Error: Image not loaded")
    exit()

image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# ---- Resize for display ----
display_scale = 0.3  #  adjust (0.2–0.5 works well)
display_img = cv2.resize(image, None, fx=display_scale, fy=display_scale)

clone = display_img.copy()

# Variables
center = None
radius = 50  # initial radius (in display scale)

# Mouse callback
def draw_circle(event, x, y, flags, param):
    global center
    if event == cv2.EVENT_LBUTTONDOWN:
        center = (x, y)

window_name = "Select Circle (Click + +/- to resize, Enter to confirm)"
cv2.namedWindow(window_name)

cv2.setMouseCallback(window_name, draw_circle)

while True:
    temp = clone.copy()

    if center is not None:
        cv2.circle(temp, center, radius, (0, 255, 0), 2)

    cv2.imshow(window_name, temp)

    key = cv2.waitKey(1) & 0xFF

    if key == ord('+') or key == ord('='):
        radius += 5

    elif key == ord('-'):
        radius = max(5, radius - 5)

    elif key == 13:  # Enter
        break

    elif key == 27:  # ESC
        cv2.destroyAllWindows()
        exit()

cv2.destroyAllWindows()

# ---- Convert back to original scale ----
x_scaled = int(center[0] / display_scale)
y_scaled = int(center[1] / display_scale)
r_scaled = int(radius / display_scale)



print("Center (original):", (x_scaled, y_scaled))
print("Radius (original):", r_scaled)
```
